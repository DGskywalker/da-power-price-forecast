# Exploratory Data Analysis: Day-Ahead Electricity Prices (DE-LU & GB)
This notebook conducts light exploratory data analysis on historical day-ahead power prices, renewable generation profiles, and weather fundamentals for the **German-Luxembourgish (DE-LU)** and **Great British (GB)** power markets.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set project paths
sys.path.insert(0, str(Path.cwd().parent / "src"))
from dappf.config import load_config

cfg = load_config("../config")
print(f"Active zones: {list(cfg.zones.keys())}")
```

## 1. Load Processed Feature Data

```python
df_de = pd.read_parquet("../data/processed/features_DE_LU.parquet")
df_gb = pd.read_parquet("../data/processed/features_GB.parquet")

print("DE-LU dataset shape:", df_de.shape)
print("GB dataset shape:", df_gb.shape)
```

## 2. Diurnal Price Profiles: Weekdays vs Weekends

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.lineplot(data=df_de, x="hour", y="price_eur_mwh", hue="is_weekend", ax=axes[0])
axes[0].set_title("DE-LU: Hourly Price Profile (Weekday vs Weekend)")
axes[0].set_ylabel("EUR / MWh")

sns.lineplot(data=df_gb, x="hour", y="price_gbp_mwh", hue="is_weekend", ax=axes[1])
axes[1].set_title("GB: Hourly Price Profile (Weekday vs Weekend)")
axes[1].set_ylabel("GBP / MWh")

plt.tight_layout()
plt.show()
```

## 3. Merit-Order Impact: Renewable Penetration vs Power Prices

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df_de.sample(2000), x="renewable_penetration_ratio", y="price_eur_mwh", alpha=0.4, color="#2b5c8f")
plt.title("DE-LU: Merit Order Effect (Renewable Penetration vs Day-Ahead Price)")
plt.xlabel("Renewable Penetration Ratio (VRE / Load)")
plt.ylabel("Price (EUR/MWh)")
plt.axhline(0, color="red", linestyle="--", alpha=0.7)
plt.show()
```
